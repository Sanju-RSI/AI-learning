```python
# token counting and cost estimations

import tiktoken
text = "I'm learning the LLM and AI"

tokens = tiktoken.encoding_for_model("gpt-4o").encode(text)

print("Text", text)
print("encoded text",tokens)
print("lenght of the tokens", len(tokens))


print("Cost Estimation")
input_tokens = len(tokens) # output tokens
output_tokens =  500
input_price_per_million = 0.50 # need to set based on the API cost for model
output_price_per_millon = 1.50 # need to set based on the API cost for model

cost = (input_tokens * input_price_per_million + output_tokens * output_price_per_millon)/1000000
print("cost are:", cost)
```

```python
import ollama
response = ollama.chat(model= "llama3.2", messages = [{"role": "user", "content": "This is user end input deviced"}])
response
```

```python
# temperature with 0 and 1
for temp in [0,1]:
    response = ollama.chat(
        model = "llama3.2",
        messages = [{"role": "user", "content": "I love coding"}],
        options = {"temperature": temp}
    )
    print("Temperature",temp)
    print("message", response["message"]["content"])
```

```python
question = "What is the refund policy of ABC Company?"

context = """
ABC Company allows refunds within 14 days
of purchase with a valid receipt.
"""

prompt = f"""
Answer the question using ONLY the context below.
If information is unavailable, say you don't know.

Context:
{context}

Question:
{question}
"""

response = ollama.chat(
    model="llama3.2",
    messages=[{"role": "user", "content": prompt}]
)

print(response["message"]["content"])
```
